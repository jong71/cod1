# 2d Array Indexing



*   1d list >> only rows
*   2d list >> rows plus column



```python
list_1d = [1,2,3,4,5,6,7,8,9]
list2d = [[1,2,3],
          [4,5,6],
          [7,8,9]]
print(list2d[1][1]) #first box is for row, the second one is for column
print(list2d[1][2])
```

```python
list_2d = [[10,20,30], [40,50,60],[70,80,90]]
arr1 = np.array(list_2d)
print(arr1)
```

```python
# [][]
# [,] > [Row, column]
```

```python
print(arr1[0])
```

```python
import numpy as np
arr = np.array([
    [0, 1, 2, 3, 4, 5],
    [10, 11, 12, 13, 14, 15],
    [20, 21, 22, 23, 24, 25],
    [30, 31, 32, 33, 34, 35],
    [40, 41, 42, 43, 44, 45],
    [50, 51, 52, 53, 54, 55],
])
```

```python
#orange
print(arr[0,3:5])
```

```python
# red
print(arr[:,2])
```

```python
# Blue
print(arr[4:,4:])
```

```python
#green "::2"
print(arr[2::2,::2])
```
